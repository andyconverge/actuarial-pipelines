```python
import pandas as pd
import numpy as np
from google.cloud import bigquery
from google.oauth2 import service_account

# =========================================================
# 1) manual input set month below, check existing database before pulling new data
# =========================================================
CREDS = "converge-database-4b8cc11a5506.json"
PROJECT_ID = "converge-database"

credentials = service_account.Credentials.from_service_account_file(CREDS)
client = bigquery.Client(credentials=credentials, project=PROJECT_ID)
```

```python


STD_DENALI_TABLE = "Data_Quality_FIA.Std_denali_trans" # target table
STD_TETON_TABLE  = "Data_Quality_FIA.Std_teton_trans"  # target table

TARGET_TABLES = [STD_DENALI_TABLE, STD_TETON_TABLE]
WRITE_MODE = "append"   # "replace" or "append"



def get_next_set_month(yyyymm: str) -> str:
    year = int(yyyymm[:4])
    month = int(yyyymm[4:6])

    if month == 12:
        return f"{year + 1}01"
    else:
        return f"{year}{month + 1:02d}"


# 1. get max existing set_month from target table
max_month_sql = f"""
SELECT MAX(set_month) AS max_set_month
FROM `{TARGET_TABLE}`
"""

max_month_df = client.query(max_month_sql).to_dataframe()
max_set_month = max_month_df.loc[0, "max_set_month"]

# 2. determine pull start month
if pd.isna(max_set_month):
    pull_from_month = None
else:
    max_set_month = str(max_set_month)
    pull_from_month = get_next_set_month(max_set_month)

# 3. build source query
if pull_from_month is None:
    source_sql = f"""
    SELECT *
    FROM `{SOURCE_TABLE}`
    """
    print("Target table is empty. Pulling full history.")
else:
    source_sql = f"""
    SELECT *
    FROM `{SOURCE_TABLE}`
    WHERE set_month >= '{pull_from_month}'
    """
    print(f"Pulling incremental data from set_month >= {pull_from_month}")

# 4. query source data
df_new = client.query(source_sql).to_dataframe()

# 5. append to target
if not df_new.empty:
    df_new.to_gbq(
        destination_table=TARGET_TABLE.replace(f"{PROJECT_ID}.", ""),
        project_id=PROJECT_ID,
        if_exists="append"
    )
    print(f"Inserted {len(df_new)} rows into {TARGET_TABLE}")
else:
    print("No new rows found.")
```

```python


# =========================================================
# 3) QUERY
#    All source tables are from denali schema
# =========================================================
query = """
WITH ser AS (
    SELECT
        policynumber AS policy_number,
        set_month AS set_month,
        totalpolicyav AS total_policy_av,
        converge AS quota_share
    FROM `denali.seriatim_values`
    WHERE set_month = @set_month
),

pol AS (
    SELECT
        policynumber AS policy_number,
        product AS product, 
        issueyear AS issue_year,
        issuemonth AS issue_month,
        issueday AS issue_day,
        CAST(annuitant_issue_age AS INTEGER) AS join_annuitant_issue_age,
        ownergender AS gender,
        annuitant_gender AS joint_annuitant_gender,
        issueage AS issue_age,
        CAST(plan AS STRING) AS term
    FROM `denali.policy`
),

prem AS (
    SELECT
        policynumber AS policy_number,
        totalinitpremium AS initial_premium
    FROM `denali.premiums`
    GROUP BY 1
)

SELECT
    s.policy_number,
    s.set_month,
    p.product,
    s.total_policy_av,
    s.quota_share,
    p.issue_year,
    p.issue_month,
    p.issue_day,
    p.join_annuitant_issue_age,
    p.gender,
    p.joint_annuitant_gender,
    p.issue_age,
    p.term,
    pr.initial_premium
FROM ser s
LEFT JOIN pol p
    ON s.policy_number = p.policy_number
LEFT JOIN prem pr
    ON s.policy_number = pr.policy_number
"""

job_config = bigquery.QueryJobConfig(
    query_parameters=[
        bigquery.ScalarQueryParameter("set_month", "STRING", SET_MONTH)
    ]
)

df_std = client.query(query, job_config=job_config).to_dataframe()

# =========================================================
# 4) CLEAN / STANDARDIZE
# =========================================================
df_std["policy_number"] = df_std["policy_number"].astype(str).str.strip()
df_std["set_month"] = df_std["set_month"].astype(str).str.strip()
df_std["product"] = df_std["product"].astype(str).str.strip().str.upper()

numeric_cols = [
    "total_policy_av",
    "quota_share",
    "issue_year",
    "issue_month",
    "issue_day",
    "join_annuitant_issue_age",
    "gender",
    "joint_annuitant_gender",
    "issue_age",
    "term",
    "initial_premium"
]

for col in numeric_cols:
    df_std[col] = pd.to_numeric(df_std[col], errors="coerce")

df_std["issue_date"] = pd.to_datetime(
    dict(
        year=df_std["issue_year"],
        month=df_std["issue_month"],
        day=df_std["issue_day"]
    ),
    errors="coerce"
)



```
